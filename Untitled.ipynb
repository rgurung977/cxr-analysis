# Does each subfolder represent one label per image?

```python
import os
import pandas as pd

# -----------------------------
# STEP 1: Give your file paths
# -----------------------------
base_folder = "files"
csv_file = "xray_chest.csv"

# Change these column names to match your CSV
image_column = "dcm"   # example: image filename or path
label_column = "type"        # example: disease label column
```

```python
# -----------------------------
# STEP 2: Read the CSV file
# -----------------------------
df = pd.read_csv(csv_file)

print("CSV loaded successfully.")
print("Number of rows in CSV:", len(df))
print("\nCSV columns are:")
print(df.columns.tolist())
```

```python
# -----------------------------
# STEP 3: Collect image files from subfolders
# -----------------------------
valid_extensions = (".jpg", ".jpeg", ".png", ".dcm")

folder_data = []   # will store filename, folder label, full path

for root, dirs, files in os.walk(base_folder):
    for file in files:
        if file.lower().endswith(valid_extensions):
            full_path = os.path.join(root, file)

            # Take the immediate folder name as label
            folder_label = os.path.basename(os.path.dirname(full_path))

            folder_data.append({
                "image_name": file,
                "folder_label": folder_label,
                "full_path": full_path
            })

folder_df = pd.DataFrame(folder_data)

print("\nTotal image files found in folders:", len(folder_df))
print("Unique folder labels found:", folder_df["folder_label"].nunique())
print("Folder labels are:", folder_df["folder_label"].unique())
print(folder_df)
```

```python
# -----------------------------
# STEP 4: Check if one image file appears in more than one folder
# -----------------------------
duplicate_images = folder_df["image_name"].duplicated()
total_duplicate_images = duplicate_images.sum()
print("\nNumber of duplicate image filenames found in folders:", total_duplicate_images)

if total_duplicate_images > 0:
    print("Warning: Same image filename may exist in multiple folders.")
    print("Matching only by image name may cause confusion.")
```

```python
# -----------------------------
# STEP 5: Check if CSV has label column
# -----------------------------
if label_column in df.columns:
    print(f"\nCSV has a label column: '{label_column}'")
    print("Unique labels in CSV:")
    print(df[label_column].dropna().unique())
else:
    print(f"\nCSV does NOT have a label column named '{label_column}'")
    print("In that case, folder names may be your only labels.")
```

```python
# -----------------------------
# STEP 6: Match folder labels with CSV labels
# -----------------------------
if image_column in df.columns:
    # Keep only needed CSV columns
    csv_small = df[[image_column]].copy()

    if label_column in df.columns:
        csv_small[label_column] = df[label_column]

    # Merge folder info with CSV using image name
    merged = pd.merge(
        folder_df,
        csv_small,
        left_on="image_name",
        right_on=image_column,
        how="left"
    )

    print("\nNumber of matched rows after merge:", len(merged))

    # Show images missing from CSV
    missing_in_csv = merged[image_column].isna().sum()
    print("Images found in folders but not found in CSV:", missing_in_csv)

    if label_column in df.columns:
        # Compare folder label with CSV label
        merged["label_match"] = merged["folder_label"].astype(str).str.lower() == merged[label_column].astype(str).str.lower()

        match_count = merged["label_match"].sum()
        mismatch_count = (~merged["label_match"]).sum()

        print("\nFolder label and CSV label comparison:")
        print("Matching labels:", match_count)
        print("Mismatching labels:", mismatch_count)

        # Show some mismatches
        mismatches = merged[merged["label_match"] == False]

        if len(mismatches) > 0:
            print("\nSome mismatched examples:")
            print(mismatches[["image_name", "folder_label", label_column]].head(10))
        else:
            print("\nAll matched rows have same folder label and CSV label.")
else:
    print(f"\nCSV does NOT have an image column named '{image_column}'")
    print("So automatic comparison with folder images cannot be done.")
```

```python
# -----------------------------
# STEP 7: Final interpretation
# -----------------------------
print("\n--- INTERPRETATION ---")

if label_column in df.columns:
    print("Since CSV has label information, the CSV should usually be treated as the source of truth.")
    print("If folder labels and CSV labels match, then each subfolder likely represents one label.")
    print("If they do not match, do not rely only on folder names.")
else:
    print("Since CSV does not have label information, each subfolder may be acting as one label.")
```

```python
import os
import pandas as pd

# 1. Parent folder containing disease subfolders
base_dir = "files"

# 2. CSV file path
csv_path = "xray_chest.csv"

# 3. Read CSV
df = pd.read_csv(csv_path)

# Change this column name to your actual CSV image path column
csv_column = "dcm"

# -----------------------------
# Images listed in CSV
# -----------------------------
csv_images = set()

for path in df[csv_column]:
    filename = os.path.basename(path)
    csv_images.add(filename)

# -----------------------------
# Images present in folders
# -----------------------------
folder_images = set()

for root, dirs, files in os.walk(base_dir):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png", ".dcm")):
            folder_images.add(file)

# -----------------------------
# Compare both
# -----------------------------

# In CSV but missing from folders
missing_from_folder = csv_images - folder_images

# In folders but missing from CSV
missing_from_csv = folder_images - csv_images

# In folders and csv

common_images = folder_images & csv_images

print("Images listed in CSV:", len(csv_images))
print("Images present in folders:", len(folder_images))

print("Images in CSV but not found in folders:", len(missing_from_folder))
print("Images in folders but not found in CSV:", len(missing_from_csv))
print("Images in folders and csv :", len(common_images)) 
```

# ***Validate and filter the DataFrame***

```python
import os
import pandas as pd

df = pd.read_csv('xray_chest.csv')

# Assume you have a base directory where images are stored
base_dir = "files"

# Create a column with the full absolute path
df['full_dcm_path'] = df['dcm'].apply(lambda x: os.path.join(base_dir, x.lstrip('/')))

# Check which files actually exist
df['dcm_exists'] = df['full_dcm_path'].apply(os.path.exists)

# Keep only rows where the DICOM file exists
df_valid = df[df['dcm_exists']].copy()

# Optionally also check JPG files
df['full_jpg_path'] = df['jpg'].apply(lambda x: os.path.join(base_dir, x.lstrip('/')))

df['jpg_exists'] = df['full_jpg_path'].apply(os.path.exists)

df_valid = df[(df['dcm_exists']) & (df['jpg_exists'])].copy()
```

```python
!pip install pydicom 
import pandas as pd
import matplotlib.pyplot as plt
import pydicom
from PIL import Image
import os

# Assuming df_valid contains columns 'full_dcm_path' and 'type' (label)
sample = df_valid.sample(5)

fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for ax, (idx, row) in zip(axes, sample.iterrows()):
    # Load DICOM
    dcm = pydicom.dcmread(row['full_dcm_path'])
    image = dcm.pixel_array
    
    # Display
    ax.imshow(image, cmap='gray')
    ax.set_title(row['type'], fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
