```python
# code to get all images

import os # helps to work with all folders and files

base_dir = "files"

all_images = [] # empty list to store image paths
cxr_folders = []
for root, dirs, files in os.walk(base_dir):
    # root = current folder path
    # files = list of files inside that folder

    for file in files:

        if file.endswith(".dcm") or file.endswith(".jpg"):

            full_path = os.path.join(root, file)

            all_images.append(full_path)

print("total images in folders:", len(all_images))
```
